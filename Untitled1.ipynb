````python
import google.generativeai as genai
from sqlalchemy import create_engine, text

# 1. API Configuration
genai.configure(api_key="AIzaSyBVGUFTpZnAGF2dXGxLE1q2EzKIHPZWLJc")

# 2. Automatic Model Detection (Ye Error solve karega)
available_models = [m.name for m in genai.list_models() if 'generateContent' in m.supported_generation_methods]
model_to_use = ""

for m in available_models:
    if 'gemini-1.5-flash' in m:
        model_to_use = m
        break

if not model_to_use:
    model_to_use = available_models[0] # Jo bhi pehla model mile

print(f"✅ Using Model: {model_to_use}")
model = genai.GenerativeModel(model_to_use)

# 3. Database Setup (Cloud Memory)
engine = create_engine('sqlite:///:memory:')
with engine.connect() as conn:
    conn.execute(text("CREATE TABLE Sales (id INTEGER, item TEXT, amount INTEGER)"))
    conn.execute(text("INSERT INTO Sales VALUES (1, 'Laptop', 50000), (2, 'Mobile', 20000), (3, 'Tablet', 15000)"))
    conn.commit()
    print("✅ Database is ready!")

# 4. AI SQL Function
def ask_database(user_query):
    prompt = f"Table 'Sales' has columns [id, item, amount]. Write ONLY the SQL query for: {user_query}"
    try:
        response = model.generate_content(prompt)
        # SQL clean karna
        sql = response.text.strip().replace('```sql', '').replace('```', '').split(';')[0]

        with engine.connect() as conn:
            result = conn.execute(text(sql))
            data = result.fetchall()
            print(f"❓ Question: {user_query}")
            print(f"💻 SQL: {sql}")
            print(f"📊 Result: {data}\n")
    except Exception as e:
        print(f"❌ Error: {e}")

# --- Test ---
ask_database("Sabse sasta item dikhao")
ask_database("Total sales kitni hui?")
````

````python
import matplotlib.pyplot as plt

def show_sales_chart():
    # 1. Database se data nikalna
    with engine.connect() as conn:
        result = conn.execute(text("SELECT item, amount FROM Sales"))
        data = result.fetchall()

    items = [row[0] for row in data]
    amounts = [row[1] for row in data]

    # 2. Graph Banana
    plt.figure(figsize=(8, 5))
    plt.bar(items, amounts, color='skyblue')
    plt.xlabel('Items')
    plt.ylabel('Amount (INR)')
    plt.title('Sales Data Visualization')
    plt.show()

# Chart dikhao
show_sales_chart()
````

````python
# Interactive Chat Input
question = input("💡 Aapka sawal kya hai? (Example: Sabse mehnga item kaunsa hai?): ")

if question:
    ask_database(question)
    # Agar sales se juda sawal hai toh chart bhi dikha dete hain
    if "total" in question.lower() or "sasta" in question.lower() or "mehnga" in question.lower() or "list" in question.lower():
        show_sales_chart()
````

````python
def ask_advanced_ai(user_query):
    prompt = f"""
    Tables available:
    1. Sales (id, item, amount, seller_id)
    2. Employees (emp_id, name, salary)

    Task: Write ONLY the SQL query for: "{user_query}"
    Output only the SQL code, nothing else.
    """
    try:
        response = model.generate_content(prompt)
        sql = response.text.strip().replace('```sql', '').replace('```', '').split(';')[0]

        with engine.connect() as conn:
            # Execute query
            result = conn.execute(text(sql))

            # Agar SELECT query hai toh data dikhao
            if result.returns_rows:
                data = result.fetchall()
                print(f"❓ Question: {user_query}")
                print(f"💻 SQL: {sql}")
                print(f"📊 Result: {data}\n")
            else:
                # Agar sirf update/insert hai toh commit karke message do
                conn.commit()
                print(f"✅ Query executed successfully (No rows to show).")

    except Exception as e:
        print(f"❌ Error: {e}")
````
